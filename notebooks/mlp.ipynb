```python
# Setup

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
```

```python
# Data
# Read data/train.csv with pandas using Id as the index
df = pd.read_csv("../data/train.csv", index_col="Id")

# Assert shape is 1460 X 80
assert df.shape == (1460, 80), f"Unexpected shape: {df.shape}"

# cast mssubclass to string
df["MSSubClass"] = df["MSSubClass"].astype(str)

print(df["SalePrice"].describe())

# Split validation and train data 80/20
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)
assert train_df.shape[0] == 1022, f"Unexpected train shape: {train_df.shape[0]}"
assert val_df.shape[0] == 219, f"Unexpected val shape: {val_df.shape[0]}"
assert test_df.shape[0] == 219, f"Unexpected test shape: {test_df.shape[0]}"

# Separate features and target
X_train = train_df.drop(columns="SalePrice")
y_train = np.log1p(train_df["SalePrice"])
X_test = test_df.drop(columns="SalePrice")
y_test = np.log1p(test_df["SalePrice"])

# Assert sale price is no longer in the test features
assert "SalePrice" not in X_test.columns, "SalePrice should not be in X_test"
X_val = val_df.drop(columns="SalePrice")
y_val = np.log1p(val_df["SalePrice"])

# Assert sale price is no longer in the features
assert "SalePrice" not in X_train.columns, "SalePrice should not be in X_train"
assert "SalePrice" not in X_val.columns, "SalePrice should not be in X_val"

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_val shape: {X_val.shape}, y_val shape: {y_val.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")

# build column lists from x_train
categorical_cols = X_train.select_dtypes(exclude="number").columns.tolist()
numerical_cols = X_train.select_dtypes(include="number").columns.tolist()

assert len(categorical_cols) + len(numerical_cols) == X_train.shape[1], "Column count mismatch"

print(f"Categorical columns: {categorical_cols}")
print(f"Numerical columns: {numerical_cols}")

```

```python
# Preprocessing

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value="None")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

preprocessor = ColumnTransformer([
    ("cat", categorical_pipeline, categorical_cols),
    ("num", numerical_pipeline, numerical_cols)
])

# Fit on X_train and transform all three sets
X_train = preprocessor.fit_transform(X_train)
X_val = preprocessor.transform(X_val)
X_test = preprocessor.transform(X_test)

print(f"X_train shape after preprocessing: {X_train.shape}")
print(f"X_val shape after preprocessing: {X_val.shape}")
print(f"X_test shape after preprocessing: {X_test.shape}")

#Check for Nan
print(f"X_train has NaN: {np.isnan(X_train).any()}")
print(f"X_val has NaN: {np.isnan(X_val).any()}")
print(f"X_test has NaN: {np.isnan(X_test).any()}")
```
