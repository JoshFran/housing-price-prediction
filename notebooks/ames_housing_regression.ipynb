# Ames Housing: Residential Sale Price Prediction

**Research question:** How accurately can tabular machine-learning models predict residential sale prices from property characteristics, and which feature/model choices matter most?

This notebook compares a median-price baseline, Linear Regression, Ridge, Lasso, and Random Forest using a held-out validation split. Preprocessing is fitted only on training rows through scikit-learn pipelines. The final test file in the Kaggle competition has no `SalePrice` labels, so it is for generating a submission only; use a validation split for model evaluation.
